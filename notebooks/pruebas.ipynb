```python
from pathlib import Path
import sys

import pandas as pd

# Localizar el proyecto desde la ra?z o desde la carpeta del notebook.
project_root = next(
    folder for folder in (Path.cwd(), *Path.cwd().parents)
    if (folder / "src" / "pronostico_geomagnetico" / "exceptions.py").is_file()
)
if str(project_root / "src") not in sys.path:
    sys.path.insert(0, str(project_root / "src"))

from pronostico_geomagnetico.exceptions import OmniLoadError

path = project_root / "data" / "raw" / "omni2_1990.dat"

COLUMN_NAMES = [
    "year",
    "day_of_year",
    "hour",
    "bartels_rotation",
    "imf_spacecraft_id",
    "plasma_spacecraft_id",
    "imf_sample_count",
    "plasma_sample_count",
    "b_magnitude_mean_nt",
    "b_vector_magnitude_nt",
    "b_latitude_gse_deg",
    "b_longitude_gse_deg",
    "bx_gse_nt",
    "by_gse_nt",
    "bz_gse_nt",
    "by_gsm_nt",
    "bz_gsm_nt",
    "sigma_b_magnitude_nt",
    "sigma_b_vector_nt",
    "sigma_bx_gse_nt",
    "sigma_by_gse_nt",
    "sigma_bz_gse_nt",
    "proton_temperature_k",
    "proton_density_cm3",
    "solar_wind_speed_kms",
    "flow_longitude_deg",
    "flow_latitude_deg",
    "alpha_proton_ratio",
    "flow_pressure_npa",
    "sigma_proton_temperature_k",
    "sigma_proton_density_cm3",
    "sigma_solar_wind_speed_kms",
    "sigma_flow_longitude_deg",
    "sigma_flow_latitude_deg",
    "sigma_alpha_proton_ratio",
    "electric_field_mvm",
    "plasma_beta",
    "alfven_mach_number",
    "kp_x10",
    "sunspot_number",
    "dst_nt",
    "ae_nt",
    "proton_flux_gt_1mev",
    "proton_flux_gt_2mev",
    "proton_flux_gt_4mev",
    "proton_flux_gt_10mev",
    "proton_flux_gt_30mev",
    "proton_flux_gt_60mev",
    "magnetospheric_flux_flag",
    "ap_nt",
    "f107_sfu",
    "pcn",
    "al_nt",
    "au_nt",
    "magnetosonic_mach_number",
]


def load_omni(path: Path) -> pd.DataFrame:
    """Lee un archivo horario de OMNI y asigna nombres a sus columnas."""

    # Comprobar que el archivo existe.
    if not path.is_file():
        raise OmniLoadError(f"No existe el archivo: {path}")

    # Leer el archivo y tratar el caso de archivo vacío.
    try:
        df = pd.read_csv(path, sep=r"\s+", header=None)
    except pd.errors.EmptyDataError as error:
        raise OmniLoadError(f"El archivo está vacío: {path}") from error

    # Comprobar la cantidad de columnas antes de nombrarlas.
    if df.shape[1] != len(COLUMN_NAMES):
        raise OmniLoadError(
            f"Se esperaban {len(COLUMN_NAMES)} columnas, "
            f"pero el archivo contiene {df.shape[1]}"
        )

    df.columns = COLUMN_NAMES
    return df


# Preparar df para las comprobaciones de las ?ltimas celdas.
df = load_omni(path)
```

```python
def validate_omni(df: pd.DataFrame) -> None:
    """Comprueba que el año y la hora tengan valores numéricos enteros."""
   
    columnas_necesarias = ["year", "day_of_year", "hour"]

    for columna in columnas_necesarias:
        if columna not in df.columns:
            raise OmniValidationError(
                f"Falta la columna obligatoria: {columna}"
            )
        
    # Validar el año.
    if df["year"].isna().any():
        raise OmniValidationError(
            "El año no puede contener valores ausentes"
        )

    if not pd.api.types.is_numeric_dtype(df["year"]):
        raise OmniValidationError(
            "El año debe contener valores numéricos"
        )

    if (df["year"] % 1 != 0).any():
        raise OmniValidationError("El año debe ser un número entero")

    # Validar la hora.
    if df["hour"].isna().any():
        raise OmniValidationError(
            "La hora no puede contener valores ausentes"
        )

    if not pd.api.types.is_numeric_dtype(df["hour"]):
        raise OmniValidationError(
            "La hora debe contener valores numéricos"
        )

    if (df["hour"] % 1 != 0).any():
        raise OmniValidationError("La hora debe ser un número entero")

    if (~df["hour"].between(MIN_HOUR, MAX_HOUR)).any():
        raise OmniValidationError(
            f"La hora debe estar entre {MIN_HOUR} y {MAX_HOUR}"
        )
    if (~df["day_of_year"].between(1, 366)).any():
        raise OmniValidationError("El día del año debe estar entre 1 y 366")
    bisiesto = ((
        df["year"] % 4 == 0)
        & ((df["year"] % 100 != 0) 
        | (df["year"] % 400 == 0))
)

    if ((df["day_of_year"] == 366) & ~bisiesto).any():
        raise OmniValidationError(
            "El día 366 solo es válido en años bisiestos"
        )
    if df.duplicated(subset=["year", "day_of_year", "hour"]).any():
        raise OmniValidationError(
        "Hay registros duplicados para el mismo año, día y hora"
    )
```

```python
import pandas as pd


def calculate_metrics(df: pd.DataFrame) -> dict:
    """Calcula la disponibilidad de las variables seleccionadas."""
    columnas = [
        "bz_gsm_nt",
        "proton_density_cm3",
        "solar_wind_speed_kms",
        "kp_x10",
    ]

    completas = df[columnas].notna().all(axis=1)
    total = len(df)

    return {
        "total_filas": total,
        "filas_completas": int(completas.sum()),
        "filas_incompletas": int((~completas).sum()),
        "porcentaje_completas": (
            float(completas.mean() * 100) if total > 0 else 0.0
        ),
    }
```
